# 01 — Data Loading
## Paddy Yield Predictor

In this notebook we just load the dataset and take a first look at it.
Nothing fancy — just checking what columns we have, the shape, missing values, and duplicates.

```python
import pandas as pd
import sys
from pathlib import Path

# Add project root to path so we can import our own modules
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.logger import get_logger
log = get_logger('01_data_loading')

log.info('Starting Data Loading notebook')
```

```python
DATA_PATH = PROJECT_ROOT / 'paddydataset.csv'

try:
    df = pd.read_csv(DATA_PATH)
    df.columns = df.columns.astype(str).str.strip()
    log.info(f'Dataset loaded — shape: {df.shape}')
except FileNotFoundError:
    log.error(f'CSV file not found at {DATA_PATH}. Please check the path.')
    raise
except Exception as e:
    log.error(f'Something went wrong while loading: {e}')
    raise
```

### Basic info about the dataset

```python
print('Shape:', df.shape)
print('\nColumn names:')
print(df.columns.tolist())
```

```python
# First 5 rows
df.head()
```

```python
# Data types and null counts
df.info()
```

```python
# Quick check: missing values and duplicates
missing = df.isna().sum().sum()
dupes = df.duplicated().sum()

print(f'Total missing values : {missing}')
print(f'Duplicate rows       : {dupes}')

log.info(f'Missing values: {missing} | Duplicates: {dupes}')
```
